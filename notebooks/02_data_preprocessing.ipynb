# Data Preprocessing

This notebook prepares the predictive maintenance dataset for machine learning.

The preprocessing pipeline includes:

- Loading the dataset
- Selecting relevant features
- Separating features and target
- Encoding categorical variables
- Train-test splitting
- Handling class imbalance
- Feature scaling
- Saving the processed datasets

The target variable is `Machine failure`.

```python
import pandas as pd
import numpy as np 

from pathlib import Path

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
raw = pd.read_csv("../data/raw/ai4i2020.csv")
```

```python
raw.shape
```

```python
raw.head()
```

```python
raw.columns.tolist()
```

```python
features = [
    "Type",
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]

target = "Machine failure"
```

```python
df = raw[features + [target]].copy()

df.head()
```

```python
df.shape
df.info()
df.isnull().sum()
```

```python
X = df.drop(columns=["Machine failure"])
y = df["Machine failure"]
```

```python
X.head()
```

```python
y.value_counts()
```

```python
X = pd.get_dummies(
    X,
    columns=["Type"],
    drop_first=True,
    dtype=int
)
```

```python
X.head()
```

```python
y.info()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
print("Training set:")
print(y_train.value_counts())
print(y_train.value_counts(normalize=True))

print("\nTest set:")
print(y_test.value_counts())
print(y_test.value_counts(normalize=True))
```

```python
from sklearn.preprocessing import StandardScaler

numeric_features = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]"
]

scaler = StandardScaler()

X_train[numeric_features] = scaler.fit_transform(
    X_train[numeric_features]
)

X_test[numeric_features] = scaler.transform(
    X_test[numeric_features]
)
```

```python
X_train[numeric_features].describe()
```

```python
X_test[numeric_features].describe()
```

```python
from pathlib import Path

processed_path = Path("../data/processed")
processed_path.mkdir(parents=True, exist_ok=True)
```

```python
X_train.to_csv(processed_path / "X_train.csv", index=False)
X_test.to_csv(processed_path / "X_test.csv", index=False)

y_train.to_csv(processed_path / "y_train.csv", index=False)
y_test.to_csv(processed_path / "y_test.csv", index=False)
```

```python
print("Processed files saved successfully.")

print(list(processed_path.iterdir()))
```

```python
print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)

print("\nMissing values:")
print("X_train:", X_train.isnull().sum().sum())
print("X_test:", X_test.isnull().sum().sum())
print("y_train:", y_train.isnull().sum())
print("y_test:", y_test.isnull().sum())

print("\nTarget distribution:")
print("Training:")
print(y_train.value_counts())

print("\nTesting:")
print(y_test.value_counts())

print("\nFeature columns:")
print(X_train.columns.tolist())
```

```python
import joblib
from pathlib import Path

models_path = Path("../models")
models_path.mkdir(parents=True, exist_ok=True)

joblib.dump(
    scaler,
    models_path / "scaler.joblib"
)

print("Scaler saved successfully.")
```

# Day 2 Complete — Data Preprocessing

The equipment sensor dataset was successfully prepared for machine learning.

### Completed Tasks
- Selected relevant sensor and machine features.
- Excluded identifier and failure-mode columns to avoid target leakage.
- Separated input features (`X`) from the target (`y`).

- Standardized numerical features using `StandardScaler`.
